## 2. Jupyter 단축키 — 시험 시간을 아껴주는 손기술

| 상황 | 단축키 | 설명 |
|---|---|---|
| 셀 실행 후 다음 셀로 | **`Shift + Enter`** | 가장 많이 씀 |
| 셀 실행 (제자리) | `Ctrl + Enter` | |
| 명령 모드로 나가기 | **`Esc`** | 셀 테두리가 파란색 |
| 위/아래에 셀 추가 | 명령 모드에서 **`A`** / **`B`** | Above / Below |
| 셀 삭제 | 명령 모드에서 **`D` `D`** | D 두 번 |
| 마크다운 / 코드 셀로 바꾸기 | 명령 모드에서 **`M`** / **`Y`** | |
| 자동 완성 | 입력 중 **`Tab`** | `df.va` + Tab → `value_counts` |
| 함수 사용법 보기 | 괄호 안에서 **`Shift + Tab`** | 파라미터 이름 확인할 때 최고 |
